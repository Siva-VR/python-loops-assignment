# Task 1: Create Database

```python
# Import required libraries
import sqlite3
import random
import pandas as pd

# Create database connection
conn = sqlite3.connect("employees.db")
cursor = conn.cursor()

# Create employees table
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees(
    emp_id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary REAL,
    years_experience INTEGER,
    performance_score REAL
)
""")

# Departments list
departments = ["Engineering", "Sales", "Marketing", "HR", "Finance"]

# Insert 40 synthetic records
for i in range(1, 41):
    name = f"Employee_{i}"
    department = random.choice(departments)
    salary = random.randint(50000, 150000)
    experience = random.randint(1, 15)
    performance = round(random.uniform(1.0, 5.0), 2)

    cursor.execute("""
    INSERT INTO employees(emp_id, name, department, salary, years_experience, performance_score)
    VALUES (?, ?, ?, ?, ?, ?)
    """, (i, name, department, salary, experience, performance))

# Commit changes
conn.commit()

print("Database and table created with 40 employee records.")
```

```python
df = pd.read_sql_query("SELECT * FROM employees", conn)
df.head()
```

# Task 2: SQL Queries


# Query1
# Employees with performance ≥ 4 AND experience ≥ 3 


```python
query1 = """
SELECT name, department, salary, performance_score
FROM employees
WHERE performance_score >= 4.0
AND years_experience >= 3
ORDER BY performance_score DESC
LIMIT 15
"""

df_query1 = pd.read_sql_query(query1, conn)
df_query1
```

# Query 2

# Salary between 70,000 and 110,000 in Engineering or Sales

```python
query2 = """
SELECT *
FROM employees
WHERE salary BETWEEN 70000 AND 110000
AND (department = 'Engineering' OR department = 'Sales')
ORDER BY department, salary DESC
"""

df_query2 = pd.read_sql_query(query2, conn)
df_query2
```

# Query 3

# Employee count and average salary per department

```python
query3 = """
SELECT department,
COUNT(*) AS employee_count,
AVG(salary) AS avg_salary
FROM employees
GROUP BY department
ORDER BY avg_salary DESC
"""

df_query3 = pd.read_sql_query(query3, conn)
df_query3
```

# SQL vs Pandas Comparison



SQL and Pandas are both powerful tools for data analysis, but they use different syntax and are suited for different tasks. SQL is a query language designed for managing and analyzing structured data stored in relational databases. For example, SQL uses operators such as = for equality, AND for combining conditions, and ORDER BY for sorting results. In contrast, Pandas uses Python syntax such as == for comparisons, & for combining conditions, and .sort_values() for sorting.

SQL is generally used when data is stored in a database system such as SQLite, MySQL, or PostgreSQL. It allows efficient querying without loading all data into memory. Pandas, on the other hand, is a Python library used for in-memory data analysis and provides powerful tools like DataFrames, groupby operations, and advanced data manipulation functions.

For datasets that are too large to fit into memory, SQL is usually the better choice because the database engine processes the data directly on disk and returns only the results needed. Pandas is better suited for smaller datasets where complex transformations, visualization, and machine learning integration are required within the Python ecosystem.
